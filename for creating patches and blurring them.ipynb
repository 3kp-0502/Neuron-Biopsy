```python
import os
import cv2

# -------- SETTINGS --------
image_path = r"C:\Users\3kp05\Desktop\Neural_Biopsy-main\new semester\all fascicles\set1_Image02_40x_bf_03_fascicle_All\1.png"      # path to your image
output_folder = r"patches_set1_Image02_40x_bf_03_fascicle_All_1"           # folder to save patches
patch_size = 128
stride = 64
# --------------------------

# create output folder
os.makedirs(output_folder, exist_ok=True)

# read image
image = cv2.imread(image_path)

h, w, c = image.shape
patch_id = 0

for y in range(0, h - patch_size + 1, stride):
    for x in range(0, w - patch_size + 1, stride):

        patch = image[y:y+patch_size, x:x+patch_size]

        patch_name = f"patch_{patch_id}.png"
        patch_path = os.path.join(output_folder, patch_name)

        cv2.imwrite(patch_path, patch)
        patch_id += 1

print(f"Total patches created: {patch_id}")
print(f"Patches saved in: {output_folder}")
```

```python
import os
import cv2

input_folder = r"patches_set1_Image02_40x_bf_03_fascicle_All_1"          # folder containing images
output_folder = r"blurred_patches_set1_Image02_40x_bf_03_fascicle_All_1" # folder to save blurred images

os.makedirs(output_folder, exist_ok=True)

for file in os.listdir(input_folder):
    
    if file.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tif")):
        
        img_path = os.path.join(input_folder, file)
        img = cv2.imread(img_path)

        # Gaussian blur with parameters (9,55)
        blurred = cv2.GaussianBlur(img, (9,9), 55)

        save_path = os.path.join(output_folder, file)
        cv2.imwrite(save_path, blurred)

print("Blurred images saved in:", output_folder)
```
