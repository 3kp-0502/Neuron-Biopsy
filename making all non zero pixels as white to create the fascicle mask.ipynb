```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# 👉 folder containing your 5 images
folder_path = r"C:\Users\3kp05\Desktop\Neural_Biopsy-main\new semester\all fascicles\set2_Image_01_20x_bf_05_all"

# get all image files
image_files = [f for f in os.listdir(folder_path) 
               if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

# sort (optional, for consistent order)
image_files.sort()

# plot setup
plt.figure(figsize=(12, 6))

for i, file in enumerate(image_files):
    img_path = os.path.join(folder_path, file)
    
    # read image
    img = cv2.imread(img_path)
    
    # create mask: non-zero pixels → white
    mask = np.any(img != 0, axis=2)
    mask_img = (mask * 255).astype(np.uint8)
    
    # plot
    plt.subplot(2, 3, i+1)  # adjust grid if needed
    plt.imshow(mask_img, cmap='gray')
    plt.title(file)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
